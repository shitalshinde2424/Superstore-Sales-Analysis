```python
# keep a backup of the original dataset

df.original = df.copy()

# keep only actual sales records

sales_df = df[df["Order Date"].notna()].copy()
```

```python
# keep only rows that contain an order Data
print("Original Dataset: ",df.shape)

print("Sales Dataset: ",sales_df.shape)
```

```python
returns_df = df.loc[
    df["Row ID"].eq("Yes"),
    ["Order ID"]
].copy()

returns_df = returns_df.drop_duplicates()
returns_df["Returned"] = "Yes"

returns_df.head()
len(returns_df)
```

```python
#These commands show how many rows were extracted, how many unique order IDs remain, and whether the new column contains "Yes".
```

```python
print(df.loc[df["Row ID"].eq("Yes")].shape)
print(returns_df.shape)
print(returns_df["Returned"].value_counts())
```

```python
returns_df = returns_df.drop_duplicates()
```

```python
# Check how many extracted return IDs exist in the sales data
matched_orders = returns_df["Order ID"].isin(sales_df["Order ID"])

print("Unique returned orders:", len(returns_df))
print("Orders found in sales data:", matched_orders.sum())
print("Orders not found in sales data:", (~matched_orders).sum())
```

```python
# here we merge the datasets
df_clean = sales_df.merge(
    returns_df,
    on="Order ID",
    how="left",
    validate="many_to_one"
)

df_clean["Returned"] = df_clean["Returned"].fillna("No")

print("Cleaned dataset shape:", df_clean.shape)
print(df_clean["Returned"].value_counts())
```

```python
# Check the number of rows and columns
print("Dataset shape:", df_clean.shape)

# Check missing values
print("\nMissing values:")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])

# Check duplicate records
print("\nDuplicate rows:", df_clean.duplicated().sum())

# Check the order dates
print("\nOrder Date data type:")
print(df_clean["Order Date"].dtype)
```

```python
# Convert Order Date to datetime
df_clean["Order Date"] = pd.to_datetime(
    df_clean["Order Date"],
    errors="coerce"
)

# Convert Ship Date to datetime
df_clean["Ship Date"] = pd.to_datetime(
    df_clean["Ship Date"],
    errors="coerce"
)

# Check the data types
print(df_clean[["Order Date", "Ship Date"]].dtypes)
```

```python
#The requirement asks you to calculate aggregate sales and profit.
```

```python
Category_analysis = df_clean.groupby("Category").agg(
    Total_Sales = ("Sales","sum"),
    Total_Profit = ("Profit","sum"),
    Total_Quantity = ("Quantity","sum")
).sort_values("Total_Sales", ascending=False)

Category_analysis
```

```python
# find the best performing category
print(
    "Highest sales category:",
    Category_analysis["Total_Sales"].idxmax()
)

print(
    "Highest profit category:",
    Category_analysis["Total_Profit"].idxmax()
)
```

```python
# This compares sales and profit across regions
```

```python
segment_analysis = df_clean.groupby("Segment").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique"),
    Unique_Customers=("Customer ID", "nunique")
)

segment_analysis = segment_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

segment_analysis
```

```python
# calculate the profit margin
segment_analysis["Profit_Margin_%"] = (
    segment_analysis["Total_Profit"]
    / segment_analysis["Total_Sales"]
) * 100

segment_analysis.round(2)
```

```python
#segment with the highest profit

best_segment = segment_analysis["Total_Profit"].idxmax()

print("Most profitable customer segment:", best_segment)
print(
    "Total profit:",
    round(segment_analysis.loc[best_segment, "Total_Profit"], 2)
)
```

```python
# region-Wise sales and profit 

region_analysis = df_clean.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Orders=("Order ID", "nunique")
)

region_analysis = region_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)

region_analysis
```

```python
# region performance
print("Highest sales region:", region_analysis["Total_Sales"].idxmax())
print("Highest profit region:", region_analysis["Total_Profit"].idxmax())
```

```python
# discount analysis (calculate the average profit at each discount level)

discount_analysis = df_clean.groupby("Discount").agg(
    Total_Sales=("Sales", "sum"),
    Average_Profit=("Profit", "mean"),
    Total_Orders=("Order ID", "nunique")
)

discount_analysis.round(2)
```

```python
#Remember, the region with the highest sales is not necessarily the region with the highest profit.

most_profitable_region = region_analysis["Total_Profit"].idxmax()

print("Most profitable region:", most_profitable_region)
print(
    "Total profit:",
    region_analysis.loc[most_profitable_region, "Total_Profit"]
)
```

```python
# here we are counting orders placed by each customers 

customer_orders = df_clean.groupby("Customer ID").agg(
    Customer_Name = ("Customer Name","first"),
    Total_Orders = ("Order ID","nunique"),
    Total_Sales = ("Sales","sum"),
    Total_Profit = ("Profit","sum")
)

customer_orders = customer_orders.sort_values(
    by = "Total_Orders",
    ascending = False
)

customer_orders.head(10)
```

```python
# total number of unique customers

total_customers = df_clean["Customer ID"].nunique()

print("Total Unique Customers : ",total_customers)
```

```python
# calculate the monthly sale
monthly_sales = df_clean.groupby("Order Month").agg(
    Total_Sales=("Sales", "sum")
)

monthly_sales.head()
```

```python
monthly_sales_plot = monthly_sales.copy()

monthly_sales_plot.index = monthly_sales_plot.index.astype(str)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales_plot.index,
    monthly_sales_plot["Total_Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Year-Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()
plt.show()
```
